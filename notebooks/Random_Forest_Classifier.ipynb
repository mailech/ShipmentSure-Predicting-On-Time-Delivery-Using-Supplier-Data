```python
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectFromModel
```

```python
data = pd.read_csv("Train.csv")

print("Dataset Shape:", data.shape)

data.head()
```

```python
X = data.drop("Reached.on.Time_Y.N", axis=1)

y = data["Reached.on.Time_Y.N"]
```

```python
X_clean = pd.get_dummies(X, drop_first=True)

print("Shape after encoding:", X_clean.shape)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=50,
    random_state=42
)
```

```python
feature_selector = SelectFromModel(
    estimator=rf_model,
    threshold="median"
)
```

```python
feature_selector.fit(X_clean, y)
```

```python
feature_mask = feature_selector.get_support()

print(feature_mask)
```

```python
selected_features = X_clean.columns[feature_mask]

print("Selected Features:")
print(selected_features)
```

```python
X_selected = feature_selector.transform(X_clean)

print("New Dataset Shape:", X_selected.shape)
```
