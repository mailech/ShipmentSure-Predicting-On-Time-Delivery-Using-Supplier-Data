# Exploratory Data Analysis (EDA)
**ShipmentSure — Predicting On-Time Delivery**

EDA is about getting to know the data before doing any complex transformations.
We look at distributions, patterns, and relationships early to spot issues.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
print("Libraries loaded.")
```

```python
# Load dataset
df = pd.read_csv('../data/Train.csv')
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Quick overview
df.describe()
```

```python
# Missing value check
print("Missing values:")
print(df.isnull().sum())
```

```python
# Target class balance — sns.countplot()
plt.figure(figsize=(6, 4))
sns.countplot(x='Reached.on.Time_Y.N', data=df, palette='Set2')
plt.title('Target Distribution: On-Time Delivery')
plt.xlabel('Reached on Time (0=No, 1=Yes)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

print(df['Reached.on.Time_Y.N'].value_counts())
```

```python
# Distribution of categorical features
cat_cols = ['Warehouse_block', 'Mode_of_Shipment', 'Product_importance', 'Gender']

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
fig.suptitle('Categorical Feature Distributions', fontsize=13)

for i, col in enumerate(cat_cols):
    sns.countplot(x=col, data=df, ax=axes[i], palette='Set3')
    axes[i].set_title(col)
    axes[i].set_xlabel('')

plt.tight_layout()
plt.show()
```

```python
# Distribution of numeric features
num_cols = ['Customer_care_calls', 'Customer_rating', 'Cost_of_the_Product',
            'Prior_purchases', 'Discount_offered', 'Weight_in_gms']

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
fig.suptitle('Numeric Feature Distributions', fontsize=13)

for idx, col in enumerate(num_cols):
    ax = axes[idx // 3][idx % 3]
    sns.histplot(df[col], kde=True, ax=ax, color='steelblue')
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

```python
# Categorical features split by target
fig, axes = plt.subplots(1, 4, figsize=(18, 4))
fig.suptitle('Categorical Features by Delivery Status', fontsize=13)

for i, col in enumerate(cat_cols):
    sns.countplot(x=col, hue='Reached.on.Time_Y.N', data=df, ax=axes[i], palette='Set1')
    axes[i].set_title(col)
    axes[i].set_xlabel('')
    axes[i].legend(title='On Time', labels=['No', 'Yes'])

plt.tight_layout()
plt.show()
```
