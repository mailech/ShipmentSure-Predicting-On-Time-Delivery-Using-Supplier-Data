# Feature Selection and Principal Component Analysis (PCA)

This notebook explores the process of identifying the most important features in the ShipmentSure dataset and reducing dimensionality using PCA for visualization and modeling.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

### Data Loading and Initial Inspection
We load the dataset and check for missing values to ensure data quality.

```python
df = pd.read_csv("Train.csv")

print("Dataset Shape:", df.shape)
display(df.head())

print("\nMissing Values:")
print(df.isnull().sum())
```

### Feature Selection
We use `SelectKBest` with the `f_classif` score function to select the top 2 features that have the strongest relationship with the target variable `Reached.on.Time_Y.N`.

```python
X = df.drop("Reached.on.Time_Y.N", axis=1)
y = df["Reached.on.Time_Y.N"]
X = pd.get_dummies(X, drop_first=True)

selector = SelectKBest(score_func=f_classif, k=2)
X_selected = selector.fit_transform(X, y)
selected_features = X.columns[selector.get_support()]

print("Selected Features:")
print(selected_features)
```

### Dimensionality Reduction with PCA
After scaling the selected features, we apply Principal Component Analysis (PCA) to reduce the data into 2 principal components.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_selected)

pca = PCA(n_components=2)
S = pca.fit_transform(X_scaled)

print("Original Shape (Selected):", X_scaled.shape)
print("PCA Shape:", S.shape)
```

### Model Training and Feature Importance
We train a `RandomForestClassifier` on the PCA components and inspect the feature importance of each component.

```python
rf = RandomForestClassifier(random_state=42)
rf.fit(S, y)

print("Model Trained Successfully")
importance = rf.feature_importances_
print("Feature Importance of PCA Components:", importance)
```

### Visualization
We visualize the distribution of the two PCA components using a boxplot.

```python
pca_df = pd.DataFrame(S, columns=["PCA1","PCA2"])

plt.figure(figsize=(7,5))
sns.boxplot(data=pca_df)

plt.title("Distribution of PCA Components")
plt.xlabel("PCA Components")
plt.ylabel("Values")
plt.show()
```

### Key Insights
- **Feature Selection**: Identified the primary drivers for on-time delivery.
- **PCA**: Transformed these drivers into a lower-dimensional space while preserving variance.
- **Distribution**: The boxplot reveals the range and potential outliers in the transformed feature space.
